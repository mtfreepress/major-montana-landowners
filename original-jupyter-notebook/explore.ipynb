# Major Montana property owners mapped

Notebook for looking at Montana property owner groupigns without cluttering other analysis notebooks.

If you're looking at this on GitHub, the interactive maps below won't load properly. Try this link instead: https://nbviewer.org/github/mtfreepress/major-montana-landowners/blob/main/explore.ipynb.

```python
# Library import and config
import json
import pandas as pd
import geopandas as gpd
pd.set_option('display.max_rows', 100) # Avoids truncation of long data table results
pd.set_option('display.max_columns', 100) # Avoids truncation of wide tata table results
```

```python
# Read in data (slow)
# Parcel data from **March 9** download -- may need to update file path
parcels = gpd.read_file('cleaned/parcels-with-mtfp-groupings.shp').to_crs(epsg=4326)

# Note grouped owner field here is initially "OwnerName_" because "OwnerName_Grouped" was truncated by .shp field name length
parcels.rename(columns={'OwnerName_': 'OwnerGroup'}, inplace=True)
```

```python
# Helper functions for common data exploration tasks

# Subset of columns that are easier to look at than the full data
majorColumns = ['CountyName','OwnerGroup','OwnerName','OwnerAddre','OwnerAdd_1','OwnerAdd_2','OwnerCity','OwnerState','PropType','TotalAcres','geometry']

def parcels_by_owner(owner_name):
    # Get parcels for given owner name
    # return parcels.loc[parcels['OwnerName'] == owner_name][majorColumns] # Ungrouped name field
    return parcels.loc[parcels['OwnerGroup'] == owner_name][majorColumns] # grouped name field (grouped is truncated by file format)

def summarize_by_owner(owner_name):
    # Text summary of given owner
    parcels = parcels_by_owner(owner_name)
    print(f'## {owner_name}')
    print(f'{parcels['TotalAcres'].sum():,.0f} acres / {len(parcels)} parcels')
    print('Owned as', parcels['OwnerName'].unique())
    print('Property types', parcels['PropType'].unique())
    print('Counties', parcels['CountyName'].unique())
    
    # return parcels

def write_to_geojson(owner_name):
    # Write parcels for given owner to GeoJson file
    parcels = parcels_by_owner(owner_name)
    fileSafeName = owner_name.replace(' ','-')
    parcels.to_file('./major-owners/' + fileSafeName + '.geojson', driver="GeoJSON")

def explore_parcels_by_owner(owner_name):
    # Produce an interactive map of parcels for given owner
    parcels = parcels_by_owner(owner_name)
    return parcels.explore(tiles="OpenStreetMap")

def search_owner(string):
    return parcels.loc[parcels['OwnerName'].notnull() & parcels['OwnerName'].str.contains(string)][majorColumns]


def summarize(owner_name):
    summarize_by_owner(owner_name)
    return explore_parcels_by_owner(owner_name)
```
